# Prédiction de MEDV avec XGBoost

Ce notebook charge le dataset Boston Housing, entraîne un modèle XGBoost et évalue ses performances.

```python
# Import de Pandas pour manipuler les données tabulaires
import pandas as pd

# Import des outils de scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Import du modèle XGBoost
from xgboost import XGBRegressor

# Import de matplotlib pour les visualisations
import matplotlib.pyplot as plt
```

```python
# URL du dataset
DATA_URL = 'https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv'

# Chargement des données
housing_data = pd.read_csv(DATA_URL)

# Aperçu
print(housing_data.head())
print('Dimensions :', housing_data.shape)
```

```python
# Variable cible
y = housing_data['medv']

# Variables explicatives
X = housing_data.drop('medv', axis=1)

# Séparation train/test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
# Création du modèle XGBoost
model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

# Entraînement
model.fit(X_train, y_train)
```

```python
# Prédictions
y_pred = model.predict(X_test)

# Métriques
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print(f'MAE  : {mae:.3f}')
print(f'MSE  : {mse:.3f}')
print(f'RMSE : {rmse:.3f}')
print(f'R²   : {r2:.3f}')
```

```python
# Importance des variables
feature_importance = pd.DataFrame({
    'Variable': X.columns,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print(feature_importance)

plt.figure(figsize=(10,6))
plt.barh(feature_importance['Variable'], feature_importance['Importance'])
plt.title('Importance des variables selon XGBoost')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```
